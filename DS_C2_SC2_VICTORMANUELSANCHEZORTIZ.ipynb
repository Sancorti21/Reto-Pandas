## Lectura de archivo y upload de bibliotecas NumPy y PANDAS

```python
import pandas as pd
import numpy as np
pd.options.display.float_format = "{:,.0f}".format
country_vaccinations = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/0.msw7wcr3ajo0.81ve3ro2jjicountry_vaccinations.csv")
country_vaccinations
```

## MOSTRAR INFORMACION RELEVANTE DEL DATA FRAME

```python
country_vaccinations.info()
country_vaccinations.shape
```

## CONVERSION DE COLUMNA DATE (OBJECT) A DTYPE DATETIME64

```python
country_vaccinations['date'] = pd.to_datetime(country_vaccinations['date'])
country_vaccinations.info()
```

## Determinar la cantidad de vacunas aplicadas de cada compañia (con base en como lo reporta cada país en la columna vaccines)

```python
vacunas_cmpy = country_vaccinations.groupby("vaccines",as_index=False
)["total_vaccinations"].sum()
vacunas_cmpy
```

```python
total_acum_by_country = country_vaccinations.sort_values("date").groupby("country").last()
acum_analitico = total_acum_by_country.groupby("vaccines",as_index=False)["total_vaccinations"].sum()
acum_analitico
```

## Obtener la cantidad de vacunas aplicadas en todo el mundo

```python
vacunas_global= total_acum_by_country["total_vaccinations"].sum()
print(f"{vacunas_global:,.0f} dosis aplicadas")
```

```python
vacunas_global = acum_analitico["total_vaccinations"].sum()
print(f"{vacunas_global:,.0f} dosis aplicadas")
```

## Calcular el promedio de vacunas aplicadas por país

```python
print(f"El promedio entre todos los paises {total_acum_by_country['total_vaccinations'].mean():,.2f}")
total_acum_by_country
```

## El promedio por pais de acuerdo a la aplicacion de vacunas diarias

```python
prom_pais = (country_vaccinations.groupby("country")["daily_vaccinations"].mean().round(2))
prom_pais
```

## Determinar la cantidad de vacunas aplicadas el dia 29/01/21 en todo el mundo

```python
total_vacunas = country_vaccinations.loc[
    country_vaccinations["date"] == "2021-01-29",
    "daily_vaccinations"].sum()

print(f"Total de vacunas aplicadas en el mundo el día 29/01/2021: {total_vacunas:,.0f}")
```

## Crear un dataframe nuevo denominado conDiferencias que contenga los datos originales y una columna derivada (diferencias) con las diferencias de aplicación entre las columnas daily_vaccinations y daily_vaccinations_raw.

```python
conDiferencias = country_vaccinations.copy()
conDiferencias["diferencias"] = conDiferencias["daily_vaccinations"] - conDiferencias["daily_vaccinations_raw"]
conDiferencias
```

## Obtener el periodo de tiempo entre el registro con fecha más reciente y el registro con fecha más antigua.

```python
periodo = (country_vaccinations["date"].max() - country_vaccinations["date"].min()).days
print(f"Periodo entre el registro más antiguo y el más reciente: {periodo} días")
```

## Crear un dataframe nuevo denominado conCantidad que contenga los datos originales y una columna derivada (canVac) con la cantidad de vacunas utilizadas cada día (usar la columna vaccines y separar por el carácter , ).

```python
conCantidad = country_vaccinations.copy()
conCantidad["canVac"] = conCantidad["vaccines"].str.split(", ").str.len()
conCantidad
```

## Generar un dataframe denominado antes20 con todos los registros que se hayan realizado antes del 20 de diciembre de 2020.

```python
antes20 = country_vaccinations.query('date < "2020-12-20"')
antes20
```

## Obtener un dataframe denominado pfizer con todos los registros donde se haya utilizado la vacuna Pfizer.

```python
pfizer = country_vaccinations[country_vaccinations["vaccines"].str.contains("Pfizer")]
pfizer
```

## Almacenar los dataframes generados (conDiferencias, conCantidad, antes20 y pfizer) en un archivo de Excel denominado resultadosReto.xlsx, donde cada dataframe ocupe una hoja diferente.

```python
with pd.ExcelWriter("resultadosReto.xlsx") as excel:
    conDiferencias.to_excel(excel, sheet_name="conDiferencias", index=False)
    conCantidad.to_excel(excel, sheet_name="conCantidad", index=False)
    antes20.to_excel(excel, sheet_name="antes20", index=False)
    pfizer.to_excel(excel, sheet_name="pfizer", index=False)
```
